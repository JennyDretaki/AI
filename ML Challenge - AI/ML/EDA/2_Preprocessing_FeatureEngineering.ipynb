# Preprocessing and Feature Engineering

### Notebook Overview:
Notebook Overview:
This notebook focuses on preprocessing and feature engineering steps.
Steps include:
1. Handling missing values
2. Encoding categorical variables
3. Scaling numerical features
4. Saving the processed dataset for modeling

**Import the Libraries**

1. Numpy for: Efficient Array Operations
2. LabelEncoder for: Encoding Categorical Data


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

**Load the needed DataSets**

```python
file_path = r'..\DataSet\train_campusrecruit.csv'
train = pd.read_csv(file_path)
```

**Display First Few Rows**

```python
train.head()
```

**Drop rows with missing values**

```python
train.dropna(inplace=True)
```

### **Data Preprocessing**
The data preprocessing steps include:
1. Encoding categorical variables
2. Scaling numerical features


**Encoding Categorical Variabes**

We convert categorical variables to numeric using Label Encoding

```python
label_cols = ['gender', 'ssc_b', 'hsc_b', 'hsc_s', 'degree_t', 'workex', 'status', 'specialisation']
label_encoders = {}
for col in label_cols:
    if col in train.columns:
        le = LabelEncoder()
        train[col] = le.fit_transform(train[col])
        label_encoders[col] = le  # Save encoders for later use
       
```

```python
train.head()
```

**Scaling Numerical Features**

```python
# List of numerical columns to scale
numerical_cols = ['ssc_p', 'hsc_p', 'degree_p', 'etest_p', 'mba_p']
scaler = StandardScaler()
# Fit and transform the training data
train[numerical_cols] = scaler.fit_transform(train[numerical_cols])
print("Scaled training data:\n", train.head())
```

**Convert Data Types**

```python
train['salary'] = train['salary'].astype(float)
```

**Handle Outliers (remove rows where salary is negative)**

```python
train = train[train['salary'] >= 0]  
```

**Calculate the median salary**

```python
median_salary = train[train['salary'] > 0]['salary'].median()  
```

**Replace only the values that are 0.0 with the median**

```python
train.loc[train['salary'] == 0.0, 'salary'] = median_salary  
```

## Save the Cleaned DataSet

```python
import os

cleaned_file_path = r'DataSet\cleaned_train_campusrecruit.csv'
# Ensure the directory exists
os.makedirs(os.path.dirname(cleaned_file_path), exist_ok=True)
train.to_csv(cleaned_file_path, index=False)

print("Data cleaning completed and saved to", cleaned_file_path)
```
